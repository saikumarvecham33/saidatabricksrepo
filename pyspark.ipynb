```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df=spark.read.format("csv")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .load("/Volumes/interview_cat/silsai/pyspark/fact_bookings.csv")

display(df)


```

```python
from pyspark.sql.functions import sum as spark_sum
agg_df=df.groupby("booking_date").agg(round(spark_sum("amount"),2).alias("total_amount")).sort("booking_date")

display(agg_df)
```

```python
df_airport=spark.read.format("csv")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .load("/Volumes/interview_cat/silsai/pyspark/dim_airports.csv")

display(df_airport)
```

```python
df_booking=spark.read.format("csv")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .load("/Volumes/interview_cat/silsai/pyspark/fact_bookings.csv")

display(df_booking)
```

```python
df_passengers=spark.read.format("csv")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .load("/Volumes/interview_cat/silsai/pyspark/dim_passengers.csv")

display(df_passengers)
```

```python
df_join=df_booking.join(df_airport,df_booking.airport_id==df_airport.airport_id,how="left")\
                .select(df_booking["booking_id"].alias("booking_id"),col("amount"),col("airport_name"))

display(df_join)
```

```python
df_gr=df_join.groupby("airport_name").agg(sum("amount").alias("total_amount")).sort("total_amount" ,ascending=False).limit(5)

display(df_gr)

```

```python
from pyspark.sql.window import Window

df_run=df_booking.withColumn("running_total",sum("amount").over(Window.partitionBy(month("booking_date")).orderBy("booking_date").rowsBetween(Window.unboundedPreceding,Window.currentRow)))
                                                              
display(df_run)

```

```python
from pyspark.sql.window import Window
win=Window.partitionBy(month("booking_date")).orderBy("booking_date").rowsBetween(Window.unboundedPreceding,Window.currentRow)

df_run=df_booking.withColumn("running_total",sum("amount").over(win))
                                                              
display(df_run)
```

```python
df=df_booking.join(df_airport,df_booking["airport_id"]==df_airport["airport_id"],how="left")\
              .select(df_booking["airport_id"],col("amount"),col("airport_name"),col("city"))

df=df.groupBy("city").agg(sum("amount").alias("total_amonut"))


df=df.withColumn("tag",when(col("total_amonut") < 1000, lit("low")).when(col("total_amonut") < 20000, lit("medium")).otherwise(lit("high")))

display(df)



            
```

```python
class my_utility:
    def __init__(self):
        pass

    def aggregator(self, df, groupby_col, aggregator_col, agg_type):
        df = df.groupby(groupby_col).agg(agg_type(aggregator_col))
        return df

obj = my_utility()

df_results=obj.aggregator(df_booking,'airport_id','amount',sum)
display(df_results)
```
